# Queens Visualization Demo

This notebook shows the optional matplotlib preview pipeline after the solver core was split from notebook drawing.

Install the optional visualization dependencies first if they are not already available:

```powershell
uv pip install -e ".[viz]"
```

````python
import json
from pathlib import Path

from queens import Board, Game, solve_regions_with_stats
from helpers.visualize import draw_board, init_board_plot
````

## Load a board

The visualization helpers take a normal `Board` instance. They do not change the solver API.

````python
game_path = Path("games/test_board.json")
regions = json.loads(game_path.read_text(encoding="utf-8"))

board = Board(regions)
board.shape, board.num_groups
````

## Preview only the region map

`init_board_plot(board)` returns a matplotlib axis with the colored regions and grid lines.

````python
ax = init_board_plot(board)
ax.figure
````

## Preview current board state

`draw_board(board)` renders the region map plus placed queens and occupied cells. In notebooks, the default `cell_output=True` clears and refreshes the output area.

````python
preview_board = Board(regions)
preview_board.place((1, 1))
draw_board(preview_board)
````

## Use the solver result for a final preview

The solver still returns plain row/column dictionaries. To visualize a solved board, replay those placements into a `Board` and draw it.

````python
result = solve_regions_with_stats(regions)
result["stats"]
````

````python
solved_board = Board(regions)
for item in result["solution"]:
    solved_board.place((item["col"] + 1, item["row"] + 1))

draw_board(solved_board)
````

## Optional interactive callback

`Game.place()` can refresh the notebook preview through a callback. The solver path does not use this; it is only for manual notebook interaction.

````python
interactive_board = Board(regions)
game = Game(interactive_board, draw_callback=draw_board)

game.place((1, 1))
````
